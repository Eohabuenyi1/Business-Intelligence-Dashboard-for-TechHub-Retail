```python
df_customers = pd.read_csv('/content/TechHub_Customers.csv')
df_products = pd.read_csv('/content/TechHub_Products.csv')
df_sales = pd.read_csv('/content/TechHub_Sales_Data.csv')

print(df_customers.head(5))
print(df_products.head(5))
print(df_sales.head(5))
```

```python
# Perform left join: start with sales_data, join customers on customer_id
merged_data = pd.merge(df_sales, df_customers, on='customer_id', how='left')

# Then join products on product_id
merged_data = pd.merge(merged_data, df_products, on='product_id', how='left')

# Display the first few rows of the merged data
print(merged_data.head())
```

```python
merged_data.drop_duplicates(inplace=True)
```

```python
merged_data.head(5)
```

```python
# Here, we keep the one from Products (right table) and drop the one from Sales (left)
if 'product_category_x' in merged_data.columns and 'product_category_y' in merged_data.columns:
    # Prefer the product_category from products (usually more accurate/official)
    merged_data['product_category'] = merged_data['product_category_y']
    merged_data.drop(columns=['product_category_x', 'product_category_y'], inplace=True)
elif 'product_category' in merged_data.columns:
    # If no conflict (unlikely here), keep it
    pass

# Optional: Display the final columns
print("Final columns:")
print(merged_data.columns.tolist())
```

```python
# Save the merged DataFrame to an Excel file
output_file = 'TechHub_Merged_Data.xlsx'
merged_data.to_excel(output_file, index=False)

print(f"Merged data saved to {output_file}")
print(f"Rows: {merged_data.shape[0]}, Columns: {merged_data.shape[1]}")
```
